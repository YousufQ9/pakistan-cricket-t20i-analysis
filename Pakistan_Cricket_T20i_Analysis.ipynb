# Analysis of Pakistan Cricket (Men's) Team in T20is (2021-2024)

#### Brief Introduction to cricket for the context of this project

Cricket is a very common if not the most common sport in the Asian subcontinent. It includes two teams (of 11 players) each taking turns 'batting' and 'bowling' and for a certain number of overs (Each over is 6 balls, so the number of overs is essentially how many balls/deliveries each team will get to bat - and make as many runs as possible.) Whichever team has made more runs at the end of the game wins the game. The total number of overs each team gets to bat is dependant on the format of the game. T20i is one of the formats in which each team gets 20 overs (or 20x6=120 balls). This will be the format that is the focus of this project. I will be conducting an analysis of Pakistan's Cricket team's performance in the T20i format in the 2021-present period. (Starting 2021 since there was an overhaul of the team and new group of players were put into the team with most of them playing consistently until now.)

The analysis of the Pakistan cricket will be done in three parts with some goals in mind. The goals of this analysis project are to find out the following:
- How has Pakistan's performance been in home games compared to away games? Has the team's performance changed throughout the years?
- What type of player roles are there on the team? Are players performing as per there role and are there any takeaways that tell us how the player roles have been affecting the team's performance?
- Who has been the most valuable player that contributed to Pakistan's victories in the analysis period.

Before starting the analysis the following are general ideas that people have with regards to the 3 goals mentioned above (these ideas are taken from news articles, sports columns or cricket discussion forums mentioned in the references below):
- Teams playing at home have a significant advantage. Not just because of home crowd support which spurs the players on but also since the home team is in charge of curating the pitch (which affects the bounce, spin and speed of the ball after bouncing before the batsmen hits it) and teams normally prepare the pitch according to their strengths. $^{[1]}$
- People think Pakistan has too many 'anchors' in their batting side. (Will be explained in more detailed later in Part 2) $^{[2]}$
- Babar Azam is considered to be the best player by many and was ranked #1 by the ranking committee for a considerable time during the analysis period. $^{[3]}$

I will look to answer the questions posed in this project which should also answer whether or not the general ideas people have of Pakistan cricket are correct.

## Part 2 - Player Roles

In this part I will look to explore the roles in the Pakistan team setup and see if I can determine how effective the players are in executing their roles. The two main types of players that are in a cricket team are batsmen and bowlers. (Each type has a variant of roles which I will explain later) To start I will use two datasets 'Batting_Stats' and 'Bowling_Stats' which contain few of the main statistics that are required for the analysis. Since these datasets have alot of data that I won't need for my analysis I will drop those columns to avoid confusion and to keep the dataset from being unnecessarily large. Finally I will merge the batting and bowling datasets into 1 'stats' dataset from which I can get whatever data I require.

```python
overall_batting = pd.read_csv('Data Sets/Batting_Stats.csv')
overall_bowling = pd.read_csv('Data Sets/Bowling_Stats.csv')
# Drop the columns which are unnecessary for this analysis.
overall_batting = overall_batting.drop(columns=['NO','BF','HS','100','50','0','4s','6s'])
overall_bowling = overall_bowling.drop(columns=['Span','Mat','Mdns','BBI','4','5'])
stats = pd.merge(overall_batting,overall_bowling,how='outer',on='Player',suffixes=('_batting', '_bowling'))
display(stats.head(6))
```

Before exploring the batting roles in the team I will briefly describe them (and some metrics) for context:
There are a number of ways a batsmen's performance can be evaluated but in my analysis I will be focusing on two metrics: Batting Average and Batting Strike Rate. Batting average is the average of how many runs a batsman scores every game and the batting strike rate is how quickly the batsman scores the runs. (The fewer the deliveries faced the higher the batting strike rate) In T20i games the batsmen look to score as quickly as possible since there are a limited number of deliveries they get to face before the other team gets to bat. In an ideal case scenario a batsman would both a high batting average and a high strike rate, however, that is unlikely since scoring quickly means the batsman is trying to hit more which gives the opponents a better chance of getting him out. This leads to two types of batsmen in a team: anchors and aggressor. Anchors play more conservatively and thus have a high average but this comes at the cost of having a lower strike rate. Aggresors play more aggressively and thus have a high strike at the cost of a lower batting average. Tail-enders are normally bowlers that have a different role in the team and thus they are not expected to have neither a high batting average nor high strike rate.

I will use kmeans to perform clustering of players to classify and see the number of anchors, aggressors and tailenders. Since the 'stats' dataset includes every player that has played for Pakistan in the analysis period I will use a filter to only include players that have played atleast 10 games to avoid skewing of data. (If a player has only played 1 game he could have performed amazingly well or terribly which would result in an inaccurate analysis) I tried different number of clusters to see how the groups are clumped in the graph and decided 5 clusters is a good number to separate the types of roles and also anomalies.



```python
batting_analysis = stats[stats['Inns_batting']!='-'].copy()
batting_analysis['Inns_batting'] = batting_analysis['Inns_batting'].astype(int)
batting_analysis = batting_analysis[['Player','Inns_batting','Ave_batting','SR_batting']]
batting_analysis = batting_analysis[batting_analysis['Inns_batting']>=10]
convert_dict = {'Ave_batting': float,'SR_batting': float}
batting_analysis = batting_analysis.astype(convert_dict)


kmeans = KMeans(n_clusters=5, random_state=1)
batting_analysis['Cluster'] = kmeans.fit_predict(batting_analysis[['Ave_batting','SR_batting']])
plt.figure(figsize=(8, 6))
for cluster in batting_analysis['Cluster'].unique():
    cluster_data = batting_analysis[batting_analysis['Cluster'] == cluster]
    plt.scatter(
        cluster_data['Ave_batting'], 
        cluster_data['SR_batting'], 
        label=f'Cluster {cluster}')

plt.xlabel('Batting Average')
plt.ylabel('Batting Strike Rate')
plt.title('Clustering Batsmen by Batting Average and Batting Strike Rate')
plt.legend()
plt.show()
```

Before continuing to the bowling analysis I will again give a brief description of the bowling roles and metrics used for analysis:

Like batting, there are quite a number of ways a bowling performance can be analyzed. In my analysis I will be focusing on two metrics: Bowling economy and bowling strike rate. Bowling economy is how much runs a bowler gives in his over on average & bowling strike rate is how quickly the bowler can get the opposing team's batsman out (the number of balls per wicket). As opposed to the batting metrics, in this case a low economy and low bowling strike rate indicates a good performance. However, much like in batting, there are different types of bowlers: containment bowlers and strike bowlers. In an ideal scenario a bowler would take lots of wickets and give as few runs as possible but trying to get batsmen out has a risk of being hit more. Hence, containment bowlers normally have a low economy but high bowling strike rate and strike bowlers have a higher economy but lower bowling strike rate.

I used the same framework for this part as for the batting analysis. Using kmeans I grouped the bowlers who had played atleast 10 games into 3 categories: containment bowlers, strike bowlers and a third category which would be neither of the first two.

```python
bowling_analysis = stats[stats['Inns_bowling']!='-'].copy()
bowling_analysis['Inns_bowling'] = bowling_analysis['Inns_bowling'].astype(int)
bowling_analysis = bowling_analysis[['Player','Inns_bowling','Econ','SR_bowling']]
bowling_analysis = bowling_analysis[bowling_analysis['Inns_bowling']>=10]
convert_dict = {'Econ': float,'SR_bowling': float}
bowling_analysis = bowling_analysis.astype(convert_dict)


kmeans = KMeans(n_clusters=3, random_state=1)
bowling_analysis['Cluster'] = kmeans.fit_predict(bowling_analysis[['Econ','SR_bowling']])

plt.figure(figsize=(8, 6))
for cluster in bowling_analysis['Cluster'].unique():
    cluster_data = bowling_analysis[bowling_analysis['Cluster'] == cluster]
    plt.scatter(
        cluster_data['Econ'], 
        cluster_data['SR_bowling'], 
        label=f'Cluster {cluster}')

plt.xlabel('Bowling Economy')
plt.ylabel('Bowling Strike Rate')
plt.title('Clustering Bowlers by Bowling Economy and Bowling Strike Rate')
plt.legend()
plt.show()
```

## Part 3 - Most Valuable Player

In this part I will look to find the most valuable player of the team. To do this I will calculate the impact a player has on a match then calculate the average impact players have in the games Pakistan won. The player with the highest average impact contributing to wins will be considered the most valuable player. To start I will choose 10 players that have been playing the most - 5 batsmen and 5 bowlers who have the highest runs and highest number of wickets respectively.

```python
display(overall_batting.head())
display(overall_bowling.head())
```

```python
Babar = pd.read_csv('Data Sets/Babar.csv')
Rizwan = pd.read_csv('Data Sets/Rizwan.csv')
Iftikhar = pd.read_csv('Data Sets/Iftikhar.csv')
Fakhar = pd.read_csv('Data Sets/Fakhar.csv')
Shadab_bat = pd.read_csv('Data Sets/Shadab(Batting).csv')
Shaheen = pd.read_csv('Data Sets/Shaheen.csv')
Rauf = pd.read_csv('Data Sets/Rauf.csv')
Shadab_bowl = pd.read_csv('Data Sets/Shadab(Bowling).csv')
Nawaz = pd.read_csv('Data Sets/Nawaz.csv')
Wasim = pd.read_csv('Data Sets/Wasim.csv')
display(Babar.head())
```

After looking at the datasets I saw that every game was included even when the match did not happen or the player did not get to bowl or bat. To start I removed all such games and then converted the datatypes to integers or floats as required. I then calculated the impact of a batsman/bowler in every game they played using the below website as the basis of the formula.

The impact calculation: 
https://www.krick3r.com/2016/02/t20-performance-impact-method.html 

The formula was slightly altered since the website was calculating the average impact over a number of games while I required the impact in every game. I used a bench mark for each of the metrics to see if the player performed good or bad based on the bench mark. The bench mark for a good performance was as follows: 

 For Batsmen  
Runs Scored = 50 
Batting Strike Rate = 140 

 For Bowlers  
Economy = 7 
Wickets = 5 

I then gave weights of 0.85 and 0.75 based on the importance of the metric. The final formulas I used were as follows: 

Player Batting Impact (per game) = ((Runs Scored)/50)*0.85 + ((Batting Strike Rate)/130)*0.75 

Player Bowling Impact (per game) = (((7/Economy)*0.75) + ((Wickets/4)*0.85)))*100 

After calculating the impact of a player on every game I merged the column with the 'pak' dataset which included every game that Pakistan played. I did this in a loop to get the impact of each of the 10 players and then used a filter to only include the games that Pakistan won. Finally I calculated the average impact of each of the players and then compared them.

```python
top_batsmen = {'Babar':Babar,'Rizwan':Rizwan,'Iftikhar':Iftikhar,'Fakhar':Fakhar,'Shadab_bat':Shadab_bat}
top_bowlers = {'Shaheen':Shaheen,'Rauf':Rauf,'Shadab_bowl':Shadab_bowl,'Nawaz':Nawaz,'Wasim':Wasim}
test = pak[['Winner','Scorecard']]
#Loop 1 to calculate the impact of batsmen
for i in top_batsmen:
    df = top_batsmen[i]
    df = (df[~((df['Bat1']=='TDNB')|(df['Bat1']=='DNB')|(df['SR']=='-'))])
    convert_dict = {'Runs':int,'BF':int,'SR':float}
    df = df.astype(convert_dict)
    df['Impact'] = (((df['Runs']/50)*0.85+(df['SR']/140)*0.75)*100).apply(lambda x: min(x, 100))
    test = pd.merge(test,df[['Game','Impact']],how='outer',left_on='Scorecard',right_on='Game').drop(columns=['Game'])\
    .rename(columns={'Impact': i})
#Loop 2 to calculate the impact of bowlers
for i in top_bowlers:
    df = top_bowlers[i]
    df = (df[~((df['Overs']=='-'))])
    convert_dict = {'Runs':int,'Wkts':int,'Econ':float}
    df = df.astype(convert_dict)
    df['Impact'] = (((7/df['Econ'])*0.75+(df['Wkts']/4)*0.85)*100).apply(lambda x: min(x, 100))
    test = pd.merge(test,df[['Game','Impact']],how='outer',left_on='Scorecard',right_on='Game').drop(columns=['Game'])\
    .rename(columns={'Impact': i})
test = test[test['Winner']=='Pakistan']
top_players=['Babar','Rizwan','Iftikhar','Fakhar','Shadab_bat','Shaheen','Rauf','Shadab_bowl','Nawaz','Wasim']
d = defaultdict(list)
#Loop to calculate the average impact of each of the players
for i in top_players:
    d[i]=("%.2f" % test[i].mean(skipna=True))
d= dict(d)
print("The average impact of the top players in the games that Pakistan won are as follows:")
print(d)
print('\n')
print('The player that had the highest average impact is',max(d, key=d.get))
```
